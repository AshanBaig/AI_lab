```
import pandas as pd
df = pd.read_csv("cereal_clean.csv")
data=df
```

```
df.head()
```

```
df = df.drop(["name","mfr","type","cups","shelf"],axis=1)
```

```
df = pd.get_dummies(df,dtype=int)
```
